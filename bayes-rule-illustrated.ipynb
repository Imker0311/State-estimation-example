Quick notebook to illustrate the posterior calculation based on the prior and likelihood function, easily adjusted for different variances. 

*Generated entirely using Gemini 3.7 Flash*

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

mu_prior, var_prior = 5.0, 1.0 # Prior: p(x) = N(mu_prior, var_prior)
mu_lik, var_lik = 10.0, 2.0 # Likelihood: p(y|x) = N(mu_lik, var_lik)

# Posterior: p(x|y) via Bayes' Rule
var_post = 1.0 / (1.0 / var_prior + 1.0 / var_lik)
mu_post = var_post * (mu_prior / var_prior + mu_lik / var_lik)

x = np.linspace(0, 15, 1000)
p_prior = norm.pdf(x, mu_prior, np.sqrt(var_prior))
p_lik = norm.pdf(x, mu_lik, np.sqrt(var_lik))
p_post = norm.pdf(x, mu_post, np.sqrt(var_post))

plt.figure(figsize=(9, 5))
plt.plot(x, p_prior, label=f'Prior $p(x) = \\mathcal{{N}}({mu_prior}, {var_prior})$', color='#00008B', lw=2)
plt.fill_between(x, p_prior, color='#00008B', alpha=0.2)

plt.plot(x, p_lik, label=f'Likelihood $p(y|x) = \\mathcal{{N}}({mu_lik}, {var_lik})$', color='orange', lw=2)
plt.vlines(mu_lik, 0, 0.75, color='black', lw=2)
plt.fill_between(x, p_lik, color='orange', alpha=0.2)

plt.plot(x, p_post, label=f'Posterior $p(x|y) = \\mathcal{{N}}({mu_post:.2f}, {var_post:.2f})$', color='green', lw=2.5)
plt.fill_between(x, p_post, color='green', alpha=0.25)

plt.xlabel("$x$", fontsize=11)
plt.ylabel("Probability Density", fontsize=11)
plt.ylim(0, 0.75)
plt.yticks([])
plt.legend(fontsize=10)
plt.show()
```
